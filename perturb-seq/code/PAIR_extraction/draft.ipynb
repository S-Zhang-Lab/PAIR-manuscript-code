```python
def reverse_complement(seq):
    complement = {
        "A": "T",
        "T": "A",
        "C": "G",
        "G": "C",
        "N": "N"  # optional: handle ambiguous bases
    }
    return "".join(complement[base] for base in reversed(seq.upper()))


seq = "CAAGTAAACCCCTACCAACTGGTCGGGGTTTGAAAC"
rc_seq = reverse_complement(seq)
print(rc_seq)
```
